## Predictive Modeling
#### 1. Select features and target
For predicting happiness scores, I am selecting feature variables like GDP per capita, social support, and healthy life expectancy because these could potentially be related to happiness. This is important because in a relevant feature selection, the model will use meaningful information to make predictions.
#### 2. Standardise features
Standardisation adjusts the features so that they have a mean of 0 and a standard deviation of 1. Standardising improves model performance because some machine learning algorithms, including linear regression, are sensitive to the scale of the input features. Features like GDP and healthy life expectancy could have very different scales, for instance, GDP is likely much larger than life expectancy, and standardizing them ensures that no feature dominates due to its scale.
#### 3. Split the data
The splitting of data into training and testing sets means train the model on one part of the data (training set) and then see how it generalizes to new, unseen data (testing set). This step is crucial to prevent overfitting. If only use one set of data for both training and testing, the model might have specific patterns in the data too well and not perform well on new data.
#### 4. Train the model
The model is trained using training data, meaning it learns the relationship between the input features. This step is the core of machine learning where the model will learn from the data through patterns. Training the model allows it to understand how each feature impacts the target and create the coefficients of their relationships.
#### 5. Make predictions
Once trained, the model may be used to predict the target variable (happiness score) for test data. Making predictions on the test to see how well the model applies to new data. 
#### 6. Evaluate the model
To evaluate the model use mean squared error (MSE) and R-squared (R^2) to assess how well the model's predictions match the actual values.

MSE

Measure the average squared difference between predicted and actual values. A lower MSE indicates better performance.

R^2

Indicate how well the model explains the variability in the target. A value of 1 mean s the model explains all the varianve while 0 means it does not explain any of the variance.

#### 7. Interpret coefficients
The model coefficients represent the strength and direction of the relationship between each feature and the target variable. The intercept is the predicted value of the target when all features are 0.

### Summary
These procedures are required to ensure that your predictive model is well-built, properly assesses its performance, and gives useful insights. Standardising data enhances model performance, splitting data helps prevent overfitting, training the model allows it to learn, and evaluating the model guarantees it can make correct predictions. Interpreting the coefficients enables you to understand the relationships between the features and the target, resulting in actionable insights.

```python
# 1. Select features and target
features = df[['GDP per capita', 'Social support', 'Healthy life expectancy']]
target = df['Score']

# 2. Standardise features
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

# 3. Split the data
X_train, X_test, y_train, y_test = train_test_split(features_scaled, target, test_size=0.2, random_state=42)

# 4. Train the model
model = LinearRegression()
model.fit(X_train, y_train)

# 5. Make predictions
y_pred = model.predict(X_test)

# 6. Evaluate the model
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("R^2:", r2_score(y_test, y_pred))

# 7. Interpret coefficients
print("Model Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
```

#### MSE
0.4637 indicates that, on average, the squared errors differences between predictions and actual happiness score are small. This means the model is performing reasonably well.
#### R^2
0.5545 means the model explains 55.45% of variability in happiness score based on the features provided. the 45% left unexplained which means other features may also contribute to happiness but not included in the model.
#### Model coefficient
- GDP per capita: For every 1 unit increase in GDP per capita, the happiness score increases bu approximately 0.3048 points.
- Social support: For every 1 unit increase in social support, the happiness score increases by approximately 0.4505 points.
- Healthy life expectancy: For every 1 unit increase in healthy life expectancy, the happiness score increase by approximately 0.2898 points.
#### Intercept:
When all features are at their baseline (0 in standardised form), the Happiness Score is expected to be 5.4180. This is the starting point of the model before the features have any effect.

### Predicted vs Actual Happiness Score

```python
plt.scatter(y_test, y_pred, alpha=0.6)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='red', linestyle='--', linewidth=2)
plt.xlabel('Actual Happiness Score')
plt.ylabel('Predicted Happiness Score')
plt.title('Predicted vs Actual Happiness Scores')
plt.show()
```

The scatterplot compares the actual happiness score (x-axis) with the predicted score (y-axis). The red dashed line represents perfect predictions. The purposes of the graph:

From the scatterplot:

- Predictions are more accurate for mid-to-high happiness score.

### Residual Plot

```python
# Calculate residuals
residuals = y_test - y_pred

# Residual plot
plt.scatter(y_pred, residuals, alpha=0.6)
plt.axhline(0, color='red', linestyle='--', linewidth=2)
plt.xlabel('Predicted Happiness Score')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```

The residual plot compares the predicted happiness score (x-axis) to residuals (y-axis), which are the differences between actual and predicted values. The red dashed line represents a residual of zero, indicating no error. The purposes of the graph:
1. To assess the model's performance and detect any patterns in prediction errors.
2. To check for random distribution of residuals, ensuring assumptions of linear regression hold.

From the residual plot:
- The residuals are randomly scattered around the red dashed line, which indicates that the model captures the linear relationship well.
- No clear pattern is visible, which means the model does not systematically underpredict or overpredict values.
- The spread of residuals is fairly constant across predicted happiness score, which reassures that variance is fairly constant.
- A few residuals are further from 0, which may indicate outliers where the predictions are less accurate.
